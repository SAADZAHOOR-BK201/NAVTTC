## Tasks 17-19 - Functions, Lambda, Map, Filter, Scope

```python
print('=== 1. Basic Functions ===')

def greet(name, greeting='Hello'):
    '''Return a greeting message.'''
    return f'{greeting}, {name}!'

def factorial(n):
    '''Recursive factorial.'''
    return 1 if n <= 1 else n * factorial(n-1)

def gcd(a, b):
    while b: a, b = b, a % b
    return a

def is_prime(n):
    if n < 2: return False
    return all(n % i != 0 for i in range(2, int(n**0.5)+1))

print(greet('Ali'))
print(greet('Sara', 'Hi'))
print('factorial(6)  =', factorial(6))
print('gcd(48,18)    =', gcd(48, 18))
print('primes 1-30   =', [x for x in range(1, 31) if is_prime(x)])

print('\n=== 2. *args and **kwargs ===')

def my_sum(*args): return sum(args)
def describe(**kwargs): return ', '.join(f'{k}={v}' for k, v in kwargs.items())

print('my_sum(1..5) =', my_sum(1, 2, 3, 4, 5))
print('describe()   =', describe(name='Ali', age=22, city='Karachi'))

print('\n=== 3. Lambda, map(), filter() ===')

nums = list(range(1, 11))
print('nums       :', nums)
print('map sq     :', list(map(lambda x: x**2, nums)))
print('map double :', list(map(lambda x: x*2, nums)))
print('filter even:', list(filter(lambda x: x%2==0, nums)))
print('filter odd :', list(filter(lambda x: x%2!=0, nums)))
print('filter >5  :', list(filter(lambda x: x>5, nums)))

words = ['hello', 'world', 'python', 'AI', 'ML']
print('map upper  :', list(map(str.upper, words)))
print('len > 4    :', list(filter(lambda w: len(w)>4, words)))

print('\n=== 4. Variable Scope (LEGB) ===')

x = 'global'

def outer_fn():
    x = 'enclosing'
    def inner_fn():
        x = 'local'
        print('  inner :', x)
    inner_fn()
    print('  outer :', x)

outer_fn()
print('  global:', x)

counter = 0
def increment():
    global counter
    counter += 1

for _ in range(5): increment()
print('counter after 5 increments:', counter)
```

## Task 20 - Nested Functions, Closures, Decorators

```python
import time, functools

print('=== 1. Nested Functions ===')

def outer(msg):
    def inner():
        print('  inner says:', msg)
    inner()

outer('Hello from outer!')

def make_adder(n):
    def adder(x): return x + n
    return adder

add5 = make_adder(5)
add10 = make_adder(10)
print('add5(3)        =', add5(3))
print('add10(7)       =', add10(7))
print('add5(add10(2)) =', add5(add10(2)))

print('\n=== 2. Closures - Power Factory ===')

def make_power(exp):
    def power(base): return base ** exp
    return power

square = make_power(2)
cube   = make_power(3)
print('square(5) =', square(5))
print('cube(3)   =', cube(3))

print('\n=== 3. Decorators ===')

def timer(func):
    @functools.wraps(func)
    def wrapper(*args, **kwargs):
        t = time.perf_counter()
        result = func(*args, **kwargs)
        elapsed = (time.perf_counter() - t) * 1000
        print(f'  {func.__name__}() took {elapsed:.3f} ms')
        return result
    return wrapper

@timer
def compute_sum(n): return sum(range(n))

result = compute_sum(1_000_000)
print('result:', result)

print('\n=== 4. Memoization via Closure ===')

def memoize(func):
    cache = {}
    def wrapper(n):
        if n not in cache: cache[n] = func(n)
        return cache[n]
    return wrapper

@memoize
def fib(n): return n if n <= 1 else fib(n-1) + fib(n-2)

print('Fibonacci series:', [fib(i) for i in range(15)])
```

## Task 23 - Strings Advanced

```python
print('=== 1. Palindrome Check ===')

def is_palindrome(s):
    s = s.lower().replace(' ', '')
    return s == s[::-1]

for w in ['racecar', 'hello', 'madam', 'python', 'level', 'A man a plan a canal Panama']:
    print(f'  {w[:27]:<28} => {is_palindrome(w)}')

print('\n=== 2. Anagram Check ===')

def is_anagram(s1, s2):
    return sorted(s1.lower()) == sorted(s2.lower())

for a, b in [('listen','silent'),('hello','world'),('triangle','integral'),('dusty','study')]:
    print(f"  anagram('{a}', '{b}') = {is_anagram(a, b)}")

print('\n=== 3. Word Frequency ===')

sentence = 'the quick brown fox jumps over the lazy dog the fox is quick'
freq = {}
for w in sentence.split(): freq[w] = freq.get(w, 0) + 1
for word, cnt in sorted(freq.items(), key=lambda x: -x[1])[:6]:
    print(f'  {word:<10}: {"#"*cnt} ({cnt})')

print('\n=== 4. Caesar Cipher ===')

def caesar(text, shift):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result.append(chr((ord(ch) - base + shift) % 26 + base))
        else:
            result.append(ch)
    return ''.join(result)

msg = 'Hello, Python is Amazing!'
enc = caesar(msg, 13)
dec = caesar(enc, -13)
print('Original :', msg)
print('ROT13    :', enc)
print('Decrypted:', dec)

print('\n=== 5. More String Ops ===')
s = 'Python Programming'
print('original     :', s)
print('reversed     :', s[::-1])
print('words rev    :', ' '.join(s.split()[::-1]))
print('vowel count  :', sum(1 for c in s.lower() if c in 'aeiou'))
print('no vowels    :', ''.join(c for c in s if c.lower() not in 'aeiou'))
print('snake_case   :', '_'.join(s.lower().split()))
```

```python
print('=== Dictionary Basics ===')
d = {'name': 'Ali Ahmed', 'age': 22, 'gpa': 3.85, 'courses': ['AI','ML','DL']}
print('dict      :', d)
print('name      :', d['name'])
print('gpa       :', d.get('gpa', 'N/A'))
print('missing   :', d.get('phone', 'N/A'))

print('\n=== CRUD ===')
d['city'] = 'Karachi'
d['age']  = 23
print('after add+update:', d)
removed = d.pop('courses')
print('pop(courses)=', removed)

print('\n=== Iteration ===')
print('keys   :', list(d.keys()))
print('values :', list(d.values()))
for k, v in d.items():
    print(f'  {k:<10}: {v}')

print('\n=== Dict Comprehension ===')
sq      = {x: x**2 for x in range(1, 11)}
cubes   = {x: x**3 for x in range(1, 6)}
even_sq = {x: x**2 for x in range(1, 11) if x%2==0}
print('squares  :', sq)
print('cubes    :', cubes)
print('even sq  :', even_sq)

print('\n=== Nested Dict - Report ===')
school = {
    'Ali'  : {'math': 90, 'science': 85, 'english': 88},
    'Sara' : {'math': 95, 'science': 92, 'english': 90},
    'Ahmed': {'math': 78, 'science': 82, 'english': 75},
}
print(f'{"Name":<8} {"Math":>6} {"Sci":>6} {"Eng":>6} {"Avg":>6}')
print('-' * 36)
for name, sc in school.items():
    avg = sum(sc.values()) / len(sc)
    print(f'{name:<8} {sc["math"]:>6} {sc["science"]:>6} {sc["english"]:>6} {avg:>6.1f}')

print('\n=== Word Count ===')
text = 'to be or not to be that is the question to be'
wc = {}
for w in text.split(): wc[w] = wc.get(w, 0) + 1
print(dict(sorted(wc.items(), key=lambda x: -x[1])))
```

## Conditional Execution, Loops & Comprehensions Recap

```python
print('=== if/elif/else ===')
for score in [95, 82, 73, 61, 45]:
    if   score >= 90: grade = 'A'
    elif score >= 80: grade = 'B'
    elif score >= 70: grade = 'C'
    elif score >= 60: grade = 'D'
    else:             grade = 'F'
    print(f'  {score} => {grade}')

print('\n=== Ternary ===')
for age in [15, 18, 25]:
    status = 'Adult' if age >= 18 else 'Minor'
    print(f'  age {age} => {status}')

print('\n=== break / continue / pass ===')
print('break at 6:', end=' ')
for i in range(1, 11):
    if i == 6: break
    print(i, end=' ')
print()

print('skip evens:', end=' ')
for i in range(1, 11):
    if i % 2 == 0: continue
    print(i, end=' ')
print()

for i in range(5):
    if i == 2: pass
    else: print(f'  i={i}')

print('\n=== Comprehensions ===')
sq     = [x**2     for x in range(1, 11)]
even_q = [x**2     for x in range(1, 11) if x%2==0]
flat   = [x        for row in [[1,2],[3,4],[5,6]] for x in row]
sq_d   = {x: x**2 for x in range(1, 6)}
sq_s   = {x**2     for x in range(-3, 4)}
print('squares  :', sq)
print('even_sq  :', even_q)
print('flat     :', flat)
print('sq_dict  :', sq_d)
print('sq_set   :', sq_s)
```
